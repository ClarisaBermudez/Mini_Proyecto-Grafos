```python

from collections import deque
import heapq

# --- Tus clases optimizadas ---
class Grafo:
    def __init__(self, dirigido = False):
        self.adyacencia = {}
        self.dirigido = dirigido

    def agregar_vertice(self, v):
        if v not in self.adyacencia:
            self.adyacencia[v] = []

    def agregar_arista(self, u, v):
        self.agregar_vertice(u)
        self.agregar_vertice(v)

        self.adyacencia[u].append(v)
        if not self.dirigido:
            self.adyacencia[v].append(u)

    def vecinos(self, v):
        return self.adyacencia.get(v, [])

    def bfs(self, inicio):
        visitados = set()
        cola = deque([inicio])
        resultado = []

        visitados.add(inicio)
        while cola:
            actual = cola.popleft()
            resultado.append(actual)
            for vecino in self.vecinos(actual):
                if vecino not in visitados:
                    visitados.add(vecino)
                    cola.append(vecino)
        return resultado

    def dfs(self, inicio, visitados = None, resultado = None):
        if visitados is None:
            visitados = set()
            resultado = []

        visitados.add(inicio)
        resultado.append(inicio)

        for vecino in self.vecinos(inicio):
            if vecino not in visitados:
                self.dfs(vecino, visitados, resultado)
        return resultado

    def mostrar_grafo(self):
        for vertice, vecinos in self.adyacencia.items():
            print(f"Servidor [{vertice}] -> Conectado a: {vecinos}")

class GrafoPonderado(Grafo):
    def __init__(self, dirigido = False):
        super().__init__(dirigido)

    def agregar_arista(self, u, v, peso):
        self.agregar_vertice(u)
        self.agregar_vertice(v)

        self.adyacencia[u].append((v, peso))
        if not self.dirigido:
            self.adyacencia[v].append((u, peso))

    def vecinos(self, v):
        return [vecino for vecino, peso in self.adyacencia.get(v, [])]

    def dijkstra(self, inicio):
        distancias = {vertice: float('Infinity') for vertice in self.adyacencia}
        distancias[inicio] = 0
        pq = [(0, inicio)]

        while pq:
            distancia_actual, actual = heapq.heappop(pq)

            if distancia_actual > distancias[actual]:
                continue
            for vecino, peso in self.adyacencia.get(actual, []):
                distancia_total = distancia_actual + peso
                if distancia_total < distancias[vecino]:
                    distancias[vecino] = distancia_total
                    heapq.heappush(pq, (distancia_total, vecino))

        return distancias


# --- Interfaz de Menú Interactivo ---
red = GrafoPonderado(dirigido=False)

# Datos iniciales de prueba (Servidores y latencia en ms)
red.agregar_arista("DataCenter-Managua", "Servidor-Masaya", 12)
red.agregar_arista("Servidor-Masaya", "Servidor-Granada", 8)
red.agregar_arista("DataCenter-Managua", "Servidor-Leon", 25)
red.agregar_arista("Servidor-Leon", "Servidor-Chinandega", 15)
red.agregar_arista("Servidor-Granada", "Servidor-Rivas", 30)

while True:
    print("\n===== SIMULADOR DE RED DE SERVIDORES =====")
    print("1. Ver topología de la red")
    print("2. Agregar enlace (arista) con latencia")
    print("3. Explorar red (BFS - Por niveles)")
    print("4. Explorar red (DFS - En profundidad)")
    print("5. Calcular latencia mínima (Dijkstra)")
    print("0. Salir")

    opcion = input("Seleccione una opción: ")

    if opcion == "1":
        print("\nTopología actual de servidores:")
        red.mostrar_grafo()

    elif opcion == "2":
        u = input("Servidor de origen: ").strip()
        v = input("Servidor de destino: ").strip()
        try:
            peso = int(input("Latencia en ms (milisegundos): "))
            red.agregar_arista(u, v, peso)
            print("¡Enlace de red agregado con éxito!")
        except ValueError:
            print("Error: La latencia debe ser un número entero.")

    elif opcion == "3":
        inicio = input("Servidor inicial para BFS: ").strip()
        if inicio in red.adyacencia:
            recorrido = red.bfs(inicio)
            print(f"Recorrido BFS desde {inicio}: {' -> '.join(recorrido)}")
        else:
            print("El servidor especificado no existe en la red.")

    elif opcion == "4":
        inicio = input("Servidor inicial para DFS: ").strip()
        if inicio in red.adyacencia:
            recorrido = red.dfs(inicio)
            print(f"Recorrido DFS desde {inicio}: {' -> '.join(recorrido)}")
        else:
            print("El servidor especificado no existe en la red.")

    elif opcion == "5":
        inicio = input("Servidor origen para Dijkstra: ").strip()
        if inicio in red.adyacencia:
            distancias = red.dijkstra(inicio)
            print(f"\nLatencias mínimas desde [{inicio}]:")
            for destino, ms in distancias.items():
                tiempo_str = f"{ms} ms" if ms != float('inf') else "Sin conexión"
                print(f" -> Hacia {destino}: {tiempo_str}")
        else:
            print("El servidor de origen no existe.")

    elif opcion == "0":
        print("Cerrando simulador de red...")
        break

    else:
        print("Opción no válida. Intente de nuevo.")
```
